### Real Sentiment Analysis on Imdb preprocessed data of keras

```python
from keras.datasets import imdb
from keras import Sequential
from keras.layers import Dense, SimpleRNN, Embedding, Flatten, LSTM
```

```python
(x_train, y_train), (x_test, y_test) = imdb.load_data()
```

```python
x_train.shape
y_train[0]

len(x_train[0]) # Trim or make all the sentences of same length
```

```python
y_test[1]
```

```python
x_test.shape
```

```python
from keras.utils import pad_sequences
```

```python
x_train = pad_sequences(x_train, padding = 'post', maxlen = 50)
x_test = pad_sequences(x_test, padding = 'post', maxlen = 50)
```

```python
len(x_train[1]) == len(x_train[2])
```

```python
model = Sequential()

model.add(SimpleRNN(50, return_sequences = False))
model.add(Dense(1, activation = 'sigmoid'))

model.summary()
```

```python
model.compile(loss = 'binary_crossentropy', optimizer= 'adam', metrics= ['accuracy'])
```

```python
model.fit(x_train, y_train, epochs = 5, validation_data = (x_test, y_test))
```
