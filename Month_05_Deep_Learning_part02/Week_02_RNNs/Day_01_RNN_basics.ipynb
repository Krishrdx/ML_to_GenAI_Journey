### Why RNN is used ???

-> For Sequential data,
it is a special type of neural network,
They store past data in their memory,
-> ANNs can't.
### To predict the blank, the model must remember previous words.

Traditional neural networks assume inputs are independent.


1. RNN inputs are timestamp for every word and input feature - total count of unique word.
2. Live industry examples : Stock price forecasting,
Speech recognition,
Sentiment analysis,
DNA sequences,
Sensor data etc
3. all involve sequential information.

### Process :

1. for example, a review system: Convert your words into a vector of n dimension.
2. Shape will be (timestamp, total count of unique word)
3. The hidden layer reoccur as the input for the next layer that's why calling it recurrent NN
4. Training Parameter will increase according to nodes.
5. The Activation Function will be **tanh** by default



```python
from keras import Sequential
from keras.layers import Dense, SimpleRNN
```

```python
model = Sequential()

model.add(SimpleRNN(3, input_shape=(4,5))) # 3 nodes, max(timestamp) = 4, input feature = 5, acti func = tanh
model.add(Dense(1, activation='sigmoid')) # binary classification

model.summary()
```

```python
print(model.get_weights()[0].shape)
```
