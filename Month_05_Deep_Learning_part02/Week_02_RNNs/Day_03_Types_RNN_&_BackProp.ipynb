### Types of RNN:

based on input and Output

 One to One - Not a RNN coz no recurrent, it will be ANN or CNN
1. One to Many
2. Many to One
3. Many to Many


Many = Sequence, One = Binary

### Examples based on types->
1. One to Many : Image Captioning and Music generation
2. Many to Many : seq2seq and same length : Part of Speech tagging, Name Entity Recognition.
variable length : Machine Translation, translate into diff lang, encoders and decoders.
3. Many to One : Sentiment Analysis, Rating Predication,
