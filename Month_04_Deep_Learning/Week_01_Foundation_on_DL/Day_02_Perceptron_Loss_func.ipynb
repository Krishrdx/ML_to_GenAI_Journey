```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

Perceptron does not output probability, only a class label.

A loss function tells us:

“How wrong is the model?”

Without loss:

No learning

No weight updates

No improvement

```python
def perceptron_loss(y, z):
    return max(0, -y * z)
```

```python
class Perceptron:
    def __init__(self, lr=0.1, epochs=10):
        self.lr = lr
        self.epochs = epochs

    def fit(self, X, y):
        n_samples, n_features = X.shape
        self.w = np.zeros(n_features)
        self.b = 0

        for _ in range(self.epochs):
            for i in range(n_samples):
                z = np.dot(X[i], self.w) + self.b
                loss = max(0, -y[i] * z)

                if loss > 0:  # misclassified
                    self.w += self.lr * y[i] * X[i]
                    self.b += self.lr * y[i]

    def predict(self, X):
        return np.sign(np.dot(X, self.w) + self.b)
```

```python
X = np.array([
    [2, 3],
    [1, -1],
    [-2, -1],
    [-3, -2]
])

y = np.array([1, 1, -1, -1])

model = Perceptron()
model.fit(X, y)

print(model.predict(X))
```
