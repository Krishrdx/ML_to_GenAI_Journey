```python
import requests
import pandas as pd
```

```python
def fetch_data():
  url = "https://api.freeapi.app/api/v1/public/stocks/INFY"
  response = requests.get(url)
  data = response.json()

  # Access dictionary keys directly for boolean checks and value extraction
  if data.get("success") and data.get("data"):
    company_name = data["data"]["company_name"]
    Listing_Date = data["data"]["Listing_Date"]
    Market_Cap = data["data"]["Market_Cap"]
    StockPE = data["data"]["StockPE"]
    Dividend_Yield = data["data"]["Dividend_Yield"]
    return company_name, Listing_Date, Market_Cap, StockPE, Dividend_Yield
  else:
    raise Exception("Failed to Fetch: " + str(data))
```

```python
def main():
  try:
    company_name, Listing_Date, Market_Cap, StockPE, Dividend_Yield = fetch_data()
    print(f"Company Name: {company_name} \n Listing Name: {Listing_Date} \n Market Cap: {Market_Cap} \n StockPE: {StockPE} \n Dividend Yield: {Dividend_Yield}")
  except Exception as e:
    print(str(e));
```

```python
if __name__ == "__main__":
  main()
```
